```python
import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split
from sklearn.neighbors import KNeighborsClassifier
from sklearn.metrics import confusion_matrix, accuracy_score, precision_score, recall_score, roc_auc_score, roc_curve

# 1. Read the dataset[cite: 91]
df = pd.read_csv('diabetes.csv')
X = df.drop(columns=['Outcome'])  # or 'Outcome' depending on your column name[cite: 92, 90]
y = df['Outcome'].values          # target column[cite: 92]

# 2. Split dataset into train and test data (stratified)[cite: 93]
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2, random_state=1, stratify=y)

# 3. Build and train the KNN classifier (using optimal k=14 found via grid search or default k=3)[cite: 93, 96]
knn = KNeighborsClassifier(n_neighbors=14)
knn.fit(X_train, y_train)

# 4. Make predictions
y_pred = knn.predict(X_test)
y_prob = knn.predict_proba(X_test)[:, 1]

# 5. Compute performance metrics[cite: 90]
cm = confusion_matrix(y_test, y_pred)
acc = accuracy_score(y_test, y_pred)
error_rate = 1 - acc
precision = precision_score(y_test, y_pred)
recall = recall_score(y_test, y_pred)
roc_auc = roc_auc_score(y_test, y_prob)

print("Confusion Matrix:\n", cm)
print(f"Accuracy: {acc:.4f}")
print(f"Error Rate: {error_rate:.4f}")
print(f"Precision: {precision:.4f}")
print(f"Recall: {recall:.4f}")
print(f"ROC-AUC Score: {roc_auc:.4f}")
```
