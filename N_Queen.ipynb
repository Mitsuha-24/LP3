```python
N = 8

def is_safe(board, r, c):
    return not any(board[i][c] or 
                   (abs(i - r) == abs(board[i].index(1) - c) if 1 in board[i] else False)
                   for i in range(N) if i != r)

def solve(board, r, f_row):
    if r == N: return True
    if r == f_row: return solve(board, r + 1, f_row)
    for c in range(N):
        if is_safe(board, r, c):
            board[r][c] = 1
            if solve(board, r + 1, f_row): return True
            board[r][c] = 0
    return False

# Initialize board and place First Queen at (row 0, col 3)
board = [[0] * N for _ in range(N)]
f_row, f_col = 0, 3
board[f_row][f_col] = 1

if solve(board, 0, f_row):
    for row in board: print(row)
else:
    print("No solution")
```
